```python
import pandas as pd
```

```python
data=pd.read_csv("trends.csv")
```

```python
num = data.select_dtypes(include="number")
```

```python
Q1 = num.quantile(0.25)
Q3 = num.quantile(0.75)
```

```python
IQR = Q3 - Q1
```

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
```

```python
print("Lower bound:\n", lower_bound)
print("Upper bound:\n", upper_bound)
```

```python
outliers = data[(data['rank'] < lower_bound['rank']) | (data['rank'] > upper_bound['rank'])]
outliers
```

```python
clean_iqr = data[(data['rank'] >= lower_bound['rank']) & (data['rank'] <= upper_bound['rank'])]
```

```python
print(clean_iqr)
```

```python
year_outliers = clean_iqr[(clean_iqr['year'] < lower_bound['year']) | (clean_iqr['year'] > upper_bound['year'])]
year_outliers
```

```python
clean_iqr = clean_iqr[(clean_iqr['year'] >= lower_bound['year']) & (clean_iqr['year'] <= upper_bound['year'])]
```
