```python
import pandas as pd
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
data = {
    'Age': [25, 45, 31, 52, 22]
}
df = pd.DataFrame(data)
```

```python
print(df)
```

```python
scaler = MinMaxScaler()
scaled = scaler.fit_transform(df['Age'])
df_scaled = pd.DataFrame(scaled, columns=['Age_Scaled'])
print(df_scaled)
```
